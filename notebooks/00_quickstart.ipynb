# Quickstart — pytorch-segmentation-models-trainer

This notebook gives you the minimal working overview of the framework.

**Recommended path:**
1. Run this notebook to verify your environment.
2. Open **`01_potsdam_windowed_segmentation.ipynb`** for the full step-by-step tutorial (data download → patch generation → training).
3. Open **`02_inference_and_export.ipynb`** to run inference on a trained checkpoint.

**What this framework does:**
- Builds sliding-window patch datasets from large GeoTIFF images.
- Trains semantic segmentation models (UNet, DeepLab, …) via PyTorch Lightning + Hydra configs.
- Runs tiled inference on full scenes and exports predictions as GeoTIFF.
